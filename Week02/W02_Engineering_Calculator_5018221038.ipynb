## Task 1: Speed Unit Conversion
Program untuk mengonversi kecepatan dari km/h ke m/s.

```
# Task 1: Speed Unit Conversion (km/h -> m/s)
speed_kmh = float(input("Enter speed [km/h]: "))
speed_ms = speed_kmh / 3.6

print(f"Speed = {speed_ms:.2f} m/s")
```

## Task 2: Power Unit Conversion
Program untuk mengonversi daya dari kW ke W.

```
# Task 2: Power Unit Conversion (kW -> W)
power_kw = float(input("Power [kW]: "))
power_w = power_kw * 1000

print(f"Power = {power_w:.2f} W")
```

## Task 3: Electrical Power and Energy Calculator
Menghitung daya (P = V x I) dan energi (E = P x t).

```
# Task 3: Electrical Power and Energy Calculator
voltage = float(input("Voltage [V]: "))
current = float(input("Current [A]: "))
operating_time = float(input("Operating Time [hour]: "))

power = voltage * current
energy = power * operating_time

print("===== ELECTRICAL CALCULATION =====")
print(f"Voltage = {voltage:.2f} V")
print(f"Current = {current:.2f} A")
print(f"Power = {power:.2f} W")
print(f"Operating Time = {operating_time:.2f} hour")
print(f"Energy = {energy:.2f} Wh")
```

## Task 4: Comparison and Logical Operators
Menggunakan hasil Task 3 untuk membuat nilai Boolean berdasarkan batas daya dan tegangan.

```
# Task 4: Comparison and Logical Operators
power_limit = 200
voltage_limit = 24

is_high_power = power > power_limit
is_high_voltage = voltage > voltage_limit

print(f"High power : {is_high_power}")
print(f"High voltage : {is_high_voltage}")
print(f"Both conditions : {is_high_power and is_high_voltage}")
print(f"Any condition : {is_high_power or is_high_voltage}")
```

## Short Reflection

1. **Apa perbedaan int dan float?**  
   `int` adalah tipe data untuk bilangan bulat (tanpa angka desimal), sedangkan `float` adalah tipe data untuk bilangan pecahan/desimal. Contoh: `5` adalah `int`, sedangkan `5.0` atau `5.25` adalah `float`.

2. **Mengapa hasil input() perlu dikonversi menggunakan float() untuk perhitungan numerik?**  
   Karena fungsi `input()` di Python selalu mengembalikan nilai bertipe `string` (teks), meskipun pengguna memasukkan angka. Jika langsung digunakan dalam operasi matematika tanpa dikonversi, program akan menghasilkan error atau melakukan penggabungan string, bukan perhitungan. Oleh karena itu perlu diubah menjadi `float` (atau `int`) terlebih dahulu.

3. **Apa perbedaan operator = dan == dalam Python?**  
   Operator `=` adalah operator penugasan (assignment) yang digunakan untuk memberikan nilai ke sebuah variabel, misalnya `x = 5`. Sedangkan `==` adalah operator perbandingan (comparison) yang digunakan untuk memeriksa apakah dua nilai sama, dan menghasilkan nilai Boolean `True` atau `False`, misalnya `x == 5`.

4. **Apa fungsi operator and dan or?**  
   Operator `and` menghasilkan `True` jika **kedua** kondisi yang dibandingkan bernilai `True`. Operator `or` menghasilkan `True` jika **salah satu** dari kondisi yang dibandingkan bernilai `True`. Keduanya digunakan untuk menggabungkan beberapa kondisi logika (Boolean) menjadi satu hasil evaluasi.

5. **Apa satu kesalahan yang Anda temui saat mengerjakan tugas ini, dan bagaimana Anda memperbaikinya?**  
   Salah satu kesalahan yang umum ditemui adalah lupa mengonversi hasil `input()` menjadi `float`, sehingga muncul `TypeError` saat mencoba melakukan operasi aritmatika seperti perkalian atau pembagian pada nilai yang masih berupa string. Kesalahan ini diperbaiki dengan membungkus `input()` menggunakan `float(...)`, misalnya `voltage = float(input("Voltage [V]: "))`, sehingga nilai yang tersimpan sudah bertipe numerik dan dapat digunakan dalam perhitungan.
